# **PARA FAZER O TREINAMENTO:**

```python
from ultralytics import YOLO

model = YOLO('yolov8m.pt')

results = model.train(
    data=r'C:\\BBD1.0\\comportamento_gado_unificado.v4i.yolov8\\data.yaml',
    epochs=80,
    imgsz=640,
    batch=16,      
    device=0,
    workers=4, 
    cache=False,    
    exist_ok=True  
)
```

```python
import shutil
import os

#Caminho de onde o YOLO salvou o resultado 
origem_modelo = '/content/runs/detect/train/weights/best.pt'
origem_grafico = '/content/runs/detect/train/results.png'
origem_matriz = '/content/runs/detect/train/confusion_matrix.png'

destino = '/content/drive/MyDrive/Kaggle/Cio_Resultados_YOLO/'

#Pasta drive
os.makedirs(destino, exist_ok=True)

#Copia os arquivos
shutil.copy(origem_modelo, destino + 'cio_melhor_modelo.pt')
shutil.copy(origem_grafico, destino + 'graficos_treino.png')
shutil.copy(origem_matriz, destino + 'matriz_confusao.png')

print(f"Arquivos salvos: {destino}")
```
